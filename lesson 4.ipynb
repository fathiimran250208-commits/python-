```python
# functions.py
# Basic function — no return
def greet(name):
    print(f"Hello, {name}!")

greet("Ali")   # Hello, Ali!

# Function with return value
def add(a, b):
    return a + b

result = add(3, 5)   # result = 8
print(result)         # 8
print(add(10,20))    # 30 — use directly

# Multiple return values
def min_max(nums):
    return min(nums), max(nums)

lo, hi = min_max([3,1,9,4])
print(lo, hi)         # 1  9

# Early return for validation
def divide(a, b):
    if b == 0: return None  # guard clause
    return a / b
divide(10,2)
```

```python
# scope.py
score = 100   # GLOBAL variable

def show_score():
    print(score)    # ✅ can READ global

def set_score():
    score = 200    # creates LOCAL copy!
    print(score)    # 200 (local)

show_score()    # 100
set_score()     # 200
print(score)     # 100 — global unchanged!

# ❌ Avoid: global keyword
def add_score(pts):
    global score
    score += pts   # modifies global — risky!

# ✅ Better: pass in, return out
def add_score(current, pts):
    return current + pts

score = add_score(score, 50)  # 150 ✅
print(score)
```
