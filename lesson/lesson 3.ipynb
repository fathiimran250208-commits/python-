```python
# for_loop.py
# Loop over a list
fruits = ["mango", "durian", "rambutan"]
for fruit in fruits:
    print(f"I love {fruit}!")

# range() — generate numbers
for i in range(1, 6):      # 1 2 3 4 5
    print(i, end=" ")

# range with step
for n in range(0, 20, 5):   # 0 5 10 15
    print(n)

# enumerate — index + value
for i, fruit in enumerate(fruits, 1):
    print(f"{i}. {fruit}")

# Loop over string
for char in "Python":
    print(char, end="-")   # P-y-t-h-o-n-

# break and continue
for n in range(10):
    if n == 5: break        # stop at 5
    if n % 2 == 0: continue # skip evens
    print(n)                  # 1 3
```

```python
# while_loop.py
# Keep asking until valid input
while True:
    age_str = input("Enter age (1-120): ")
    if age_str.isdigit():
        age = int(age_str)
        if 1 <= age <= 120: break
    print("Invalid! Try again.")

# Number guessing game
secret   = 42
attempts = 0
while True:
    guess    = int(input("Guess: "))
    attempts += 1
    if   guess < secret: print("Too low 📉")
    elif guess > secret: print("Too high 📈")
    else:
        print(f"✅ Got it in {attempts} tries!")
        break

# Menu loop
while True:
    print("1) Add  2) View  3) Quit")
    choice = input("Choice: ")
    if choice == "3": break
```

```python
# lists.py
# Create a list
scores = [85, 92, 67, 78, 90]
names  = ["Ali", "Siti", "Ravi"]
mixed  = [1, "hello", True, 3.14]

# Access by index (starts at 0)
print(scores[0])   # 85 — first
print(scores[-1])  # 90 — last
print(scores[1:3]) # [92, 67] — slice

# Modify
scores.append(95)  # add to end
scores.insert(0, 70) # insert at index
scores.remove(67)  # remove first match
scores.pop()       # remove last, return it

# Useful functions
print(len(scores))  # count items
print(sum(scores))  # total
print(max(scores))  # highest
print(min(scores))  # lowest
scores.sort()       # sort in place
sorted(scores)      # returns new sorted list
```

```python
# dicts.py
# Create a dictionary
student = {
    "name":   "Ali",
    "age":    22,
    "score":  87.5,
    "passed": True
}

# Read values
print(student["name"])            # Ali
print(student.get("grade", "N/A")) # N/A (safe!)

# Update & add
student["score"] = 92.0   # update
student["grade"] = "A"    # add new key

# Delete
del student["passed"]
removed = student.pop("age")

# Loop over dict
for key, val in student.items():
    print(f"{key}: {val}")

# Dict comprehension
doubled = {k: v*2 for k,v in {"a":1,"b":2}.items()}
```

```python
# contact_book.py
import json, os

FILE = "contacts.json"
contacts = json.load(open(FILE)) if os.path.exists(FILE) else {}

def save():
    with open(FILE, "w") as f: json.dump(contacts, f, indent=2)

while True:
    print("\n1)Add 2)Search 3)List 4)Delete 5)Quit")
    c = input("Choice: ")
    if c == "1":
        n = input("Name: ")
        contacts[n] = {"phone": input("Phone: "), "email": input("Email: ")}
        save()
        print(f"✅ {n} saved!")
    elif c == "2":
        n = input("Search name: ")
        print(contacts.get(n, "Not found"))
    elif c == "3":
        for n, info in contacts.items(): print(f"{n}: {info}")
    elif c == "4":
        n = input("Delete: "); contacts.pop(n, None); save()
    elif c == "5": break
```
