```python

# Task 1: Number Games

# (a) Multiplication table
number = int(input("Enter a number for its multiplication table: "))

print(f"\nMultiplication table for {number}")
for i in range(1, 13):
    print(f"{number} x {i} = {number * i}")

# (b) Prime numbers between 1 and 100
print("\nPrime numbers between 1 and 100:")

for num in range(2, 101):
    is_prime = True

    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print(num, end=" ")

print()

# (c) FizzBuzz from 1 to 50
print("\nFizzBuzz:")

for num in range(1, 51):
    if num % 3 == 0 and num % 5 == 0:
        print("FizzBuzz")
    elif num % 3 == 0:
        print("Fizz")
    elif num % 5 == 0:
        print("Buzz")
    else:
        print(num)
```

```python

# Task 2: Shopping Cart

cart = []

while True:
    print("\n--- Shopping Cart ---")
    print("1. Add item")
    print("2. Remove item")
    print("3. Update quantity")
    print("4. View cart and total")
    print("5. Clear cart")
    print("6. Quit")

    choice = input("Choose an option: ").strip()

    if choice == "1":
        name = input("Item name: ").strip()

        try:
            price = float(input("Item price: "))
            qty = int(input("Quantity: "))

            if not name or price < 0 or qty <= 0:
                print("Enter a name, non-negative price and positive quantity.")
                continue

            cart.append({
                "name": name,
                "price": price,
                "qty": qty
            })
            print("Item added!")

        except ValueError:
            print("Please enter a valid price and whole-number quantity.")

    elif choice == "2":
        name = input("Enter item name to remove: ").strip()
        for item in cart:
            if item["name"].lower() == name.lower():
                cart.remove(item)
                print("Item removed!")
                break
        else:
            print("Item not found.")

    elif choice == "3":
        name = input("Enter item name to update: ").strip()
        for item in cart:
            if item["name"].lower() == name.lower():
                try:
                    qty = int(input("Enter new quantity: "))
                    if qty <= 0:
                        print("Quantity must be positive.")
                    else:
                        item["qty"] = qty
                        print("Quantity updated!")
                except ValueError:
                    print("Enter a valid whole number.")
                break
        else:
            print("Item not found.")

    elif choice == "4":
        total = 0
        print("\n--- Your Cart ---")
        for item in cart:
            item_total = item["price"] * item["qty"]
            total += item_total
            print(
                f"{item['name']:<15} "
                f"{item['qty']} x ${item['price']:.2f} "
                f"= ${item_total:.2f}"
            )
        print(f"Total: ${total:.2f}")

    elif choice == "5":
        cart.clear()
        print("Cart cleared!")

    elif choice == "6":
        print("Goodbye!")
        break

    else:
        print("Invalid option. Choose 1-6.")
```

```python

students = {}

while True:
    print("\n1. Add student")
    print("2. Add score")
    print("3. View report card")
    print("4. Find top student")
    print("5. Exit")

    choice = input("Enter choice: ")

    # 1. Add student
    if choice == "1":
        name = input("Enter student name: ")

        if name not in students:
            students[name] = {}
            print("Student added!")
        else:
            print("Student already exists!")

    # 2. Add score
    elif choice == "2":
        name = input("Enter student name: ")

        if name in students:
            subject = input("Enter subject: ")
            score = float(input("Enter score: "))

            students[name][subject] = score
            print("Score added!")
        else:
            print("Student not found!")

    # 3. View report card
    elif choice == "3":
        name = input("Enter student name: ")

        if name in students and students[name]:
            scores = list(students[name].values())

            print("Scores:", students[name])
            print("Average:", sum(scores) / len(scores))
            print("Highest:", max(scores))
            print("Lowest:", min(scores))
        else:
            print("Student not found or no scores!")

    # 4. Find top student overall
    elif choice == "4":
        top_student = ""
        highest_avg = -1

        for name in students:
            scores = list(students[name].values())

            if scores:
                avg = sum(scores) / len(scores)

                if avg > highest_avg:
                    highest_avg = avg
                    top_student = name

        if top_student != "":
            print("Top student:", top_student)
            print("Average:", highest_avg)
        else:
            print("No scores available!")

    # 5. Exit
    elif choice == "5":
        print("Goodbye!")
        break

    else:
        print("Invalid choice!")
```

```python
import string

def analyze_word_frequency(paragraph):
    # Define stop words as requested
    stop_words = {"a", "the", "is", "in", "of", "and", "to", "it"}

    # Use str.translate() to strip all punctuation
    # string.punctuation contains characters like ! " # $ % & ' ( ) * + , - . / : ; < = > ? @ [ \ ] ^ _ ` { | } ~
    translation_table = str.maketrans("", "", string.punctuation)
    clean_paragraph = paragraph.translate(translation_table)

    # Split into words and convert to lowercase
    words = clean_paragraph.lower().split()

    # Build frequency dictionary excluding stop words
    frequency_dict = {}
    for word in words:
        if word not in stop_words:
            frequency_dict[word] = frequency_dict.get(word, 0) + 1

    # Sort items: 
    # 1. Primary sort key: frequency in descending order (-count)
    # 2. Secondary sort key: word alphabetically in ascending order (word)
    sorted_words = sorted(frequency_dict.items(), key=lambda item: (-item[1], item[0]))

    # Print top 10 words
    print("\n--- Top 10 Words ---")
    top_10 = sorted_words[:10]
    for rank, (word, count) in enumerate(top_10, start=1):
        print(f"{rank}. {word}: {count}")

    return frequency_dict


# --- Interactive Demo ---
if __name__ == "__main__":
    sample_paragraph = """
    Data science is an amazing field. It involves statistics, programming, and data analysis. 
    The field of data science is growing rapidly in the modern world, and it offers great opportunities to everyone.
    """
    
    print("Analyzing sample paragraph...")
    analyze_word_frequency(sample_paragraph)
```
