```python
# class_basics.py
class BankAccount:
    """A simple bank account."""

    def __init__(self, owner, balance=0):
        self.owner   = owner   # attribute
        self.balance = balance # attribute

    def deposit(self, amount):
        self.balance += amount
        print(f"✅ Deposited RM{amount}")

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("Insufficient funds!")
        self.balance -= amount

    def __str__(self):
        return f"{self.owner}: RM{self.balance:.2f}"

# Create instances (objects)
ali  = BankAccount("Ali", 1000)
siti = BankAccount("Siti")

ali.deposit(500)       # RM1500
ali.withdraw(200)      # RM1300
print(ali)             # Ali: RM1300.00
```

```python
# inheritance.py
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."  # override in subclass

    def __str__(self):
        return f"{self.name} says {self.speak()}"

class Dog(Animal):    # inherits Animal
    def speak(self): return "Woof!"
    def fetch(self):  print("Fetching!")

class Cat(Animal):
    def speak(self): return "Meow!"

class Owl(Animal):
    def speak(self): return "Hoo-hoo!"

d = Dog("Rex")
c = Cat("Whiskers")
o = Owl("Harry")
print(d)          # Rex says Woof!
print(c)          # Whiskers says Meow!
print(o)          # Harry says Hoo-hoo!

# @dataclass — less boilerplate
from dataclasses import dataclass

@dataclass
class Student:
    name:  str
    age:   int
    score: float = 0.0

s = Student("Ali", 20, 85.5)
print(s)  # Student(name=Ali, age=20, score=85.5)
```

```python
import re

def palindrome(s: str) -> bool:
    """Check if a string is a palindrome, ignoring non-alphanumeric characters and case."""
    cleaned = ''.join(c.lower() for c in s if c.isalnum())
    return cleaned == cleaned[::-1]

def count_vowels(s: str) -> int:
    """Count the number of vowels (a, e, i, o, u) in a string case-insensitively."""
    vowels = "aeiouAEIOU"
    return sum(1 for c in s if c in vowels)

def title_case(s: str) -> bool:
    """Convert a string to title case without using the built-in .title() method."""
    words = s.split(' ')
    titled_words = []
    for word in words:
        if word:
            # Capitalize first character and lowercase the rest manually
            titled_word = word[0].upper() + word[1:].lower()
            titled_words.append(titled_word)
        else:
            titled_words.append('')
    return ' '.join(titled_words)

def truncate(s: str, max_len: int, suffix: str = '...') -> str:
    """Truncate a string if it exceeds max_len, appending the suffix."""
    if len(s) <= max_len:
        return s
    return s[:max_len - len(suffix)] + suffix

def slugify(s: str) -> str:
    """Convert a string to a URL-friendly slug (lowercase, spaces to '-', remove non-alphanumeric)."""
    s = s.lower()
    # Replace spaces with hyphens
    s = re.sub(r'\s+', '-', s)
    # Remove all characters that are not alphanumeric or hyphens
    s = re.sub(r'[^a-z0-9\-]', '', s)
    # Remove duplicate consecutive hyphens
    s = re.sub(r'-+', '-', s)
    return s.strip('-')
print(palindrome("madam"))
print(count_vowels("Hello World"))
print(title_case("hello WORLD python"))
print(truncate("Hello World", 5))
print(slugify("Hello World!"))
```

```python
import random
from datetime import datetime

def run_math_quiz():
    operators = ['+', '-', '*']
    score = 0
    wrong_questions = []
    total_questions = 10

    print("=== Welcome to the Random Math Quiz! ===")
    start_time = datetime.now()

    for i in range(1, total_questions + 1):
        num1 = random.randint(1, 20)
        num2 = random.randint(1, 20)
        op = random.choice(operators)
        
        # Avoid negative results for subtraction to keep it straightforward
        if op == '-' and num1 < num2:
            num1, num2 = num2, num1

        expression = f"{num1} {op} {num2}"
        correct_answer = eval(expression)

        try:
            user_answer = int(input(f"Question {i}: What is {expression}? "))
        except ValueError:
            print("Invalid input. Counted as incorrect.")
            user_answer = None

        if user_answer == correct_answer:
            print("Correct!\n")
            score += 1
        else:
            print(f"Wrong! The correct answer was {correct_answer}.\n")
            wrong_questions.append({
                "question": expression,
                "your_answer": user_answer,
                "correct_answer": correct_answer
            })

    end_time = datetime.now()
    time_taken = end_time - start_time
    percentage = (score / total_questions) * 100

    print("\n" + "="*30)
    print("QUIZ RESULTS")
    print("="*30)
    print(f"Score: {score}/{total_questions}")
    print(f"Percentage: {percentage:.2f}%")
    print(f"Time Taken: {time_taken}")

    if wrong_questions:
        print("\nQuestions answered incorrectly:")
        for w in wrong_questions:
            print(f"- {w['question']} = {w['correct_answer']} (You answered: {w['your_answer']})")
    else:
        print("\nFantastic! You got all questions correct!")

if __name__ == "__main__":
    run_math_quiz()
```

```python
import random
import string
import time
import textwrap

def generate_password(length=12, symbols=True, *args, **kwargs):
    """Generate a secure random password."""
    letters = string.ascii_letters
    digits = string.digits
    punctuation = string.punctuation if symbols else ''
    
    all_chars = letters + digits + punctuation
    if not all_chars:
        all_chars = letters

    password = ''.join(random.choice(all_chars) for _ in range(length))
    return password

def format_table(*headers, **rows):
    """Format a simple text table using headers and keyword rows."""
    # Example usage: format_table("Name", "Age", row1=["Alice", 25], row2=["Bob", 30])
    col_widths = [len(str(h)) for h in headers]
    
    row_values = list(rows.values())
    for row in row_values:
        for idx, val in enumerate(row):
            if idx < len(col_widths):
                col_widths[idx] = max(col_widths[idx], len(str(val)))
            else:
                col_widths.append(len(str(val)))

    header_line = " | ".join(str(h).ljust(col_widths[i]) for i, h in enumerate(headers))
    separator = "-+-".join("-" * w for w in col_widths)
    
    lines = [header_line, separator]
    for row in row_values:
        line = " | ".join(str(val).ljust(col_widths[i]) if i < len(col_widths) else str(val) for i, val in enumerate(row))
        lines.append(line)
        
    return "\n".join(lines)

def countdown_timer(seconds, *args, **kwargs):
    """Count down from a given number of seconds with visual output."""
    print(f"Starting countdown for {seconds} seconds...")
    while seconds > 0:
        mins, secs = divmod(seconds, 60)
        timer_display = f"{mins:02d}:{secs:02d}"
        print(timer_display, end="\r")
        time.sleep(1)
        seconds -= 1
    print("00:00 - Time's up!      ")

def caesar_cipher(text, shift=3, mode='encrypt', *args, **kwargs):
    """Encrypt or decrypt text using the Caesar cipher method."""
    result = []
    if mode == 'decrypt':
        shift = -shift

    for char in text:
        if char.isalpha():
            ascii_offset = ord('A') if char.isupper() else ord('a')
            shifted_char = chr((ord(char) - ascii_offset + shift) % 26 + ascii_offset)
            result.append(shifted_char)
        else:
            result.append(char)
            
    return "".join(result)

def word_wrap(text, width=80, *args, **kwargs):
    """Wrap text to a specified maximum line width."""
    return "\n".join(textwrap.wrap(text, width=width))

def format_currency(amount, currency='MYR', decimals=2, *args, **kwargs):
    """Format a numeric amount as currency with specified code and decimals."""
    return f"{currency} {amount:,.{decimals}f}"
# Test all functions

print("1. Password:")
print(generate_password())
print(generate_password(8, symbols=False))

print("\n2. Table:")
print(format_table("Name", "Age", row1=["Alice", 25], row2=["Bob", 30]))

print("\n3. Countdown:")
# Uncomment the next line to test the timer
# countdown_timer(5)

print("\n4. Caesar Cipher:")
print(caesar_cipher("Hello World"))
print(caesar_cipher("Hello World", shift=5))

print("\n5. Word Wrap:")
print(word_wrap(
    "Python is a beginner-friendly programming language "
    "that is widely used for many different applications.",
    width=30
))

print("\n6. Currency:")
print(format_currency(1234.5))
print(format_currency(1234.5, currency="USD"))
print(format_currency(1234.5678, decimals=3))
```
