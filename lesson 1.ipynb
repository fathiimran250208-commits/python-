```python

# casting.py
# Check the type of anything
print(type("hello"))     # <class 'str'>
print(type(16))         # <class 'int'>
print(type(3.14))       # <class 'float'>
print(type(True))        # <class 'bool'>

# int() — string/float → integer
print(type("95"))
int("95")      # → 95
print(type(int("95")))
int(3.9)       # → 3  (truncates!)
int("95")      # → 95
int(3.9)       # → 3  (truncates!)
print(int(3.9))

# float() — string/int → decimal
float("1.72")  # → 1.72
float(5)       # → 5.0

# str() — anything → text
str(100)       # → "100"
str(True)      # → "True"

# bool() — falsy values → False
bool(0)        # → False
bool("")       # → False (empty string)
bool(42)       # → True
bool("hello")  # → True
```

```python
# task1_profile_card.py

# Collect user details
name = input("Enter your name: ").strip()
age = input("Enter your age: ").strip()
country = input("Enter your country: ").strip()
colour = input("Enter your favourite colour: ").strip()
fun_fact = input("Enter a fun fact about yourself: ").strip()

# Prepare content lines
title = " PERSONAL PROFILE "
line_name = f"Name: {name}"
line_age = f"Age: {age}"
line_country = f"Country: {country}"
line_colour = f"Favourite Colour: {colour}"
line_fact = f"Fun Fact: {fun_fact}"

# Calculate dynamic width based on the longest string
max_content_length = max(
    len(title),
    len(line_name),
    len(line_age),
    len(line_country),
    len(line_colour),
    len(line_fact),
)
padding = 10
card_width = max_content_length + padding

border = "=" * card_width

# Output the formatted profile card
print("\n" + border)
print(f"{title:^{card_width}}")
print(border)
print(f"  {line_name}")
print(f"  {line_age}")
print(f"  {line_country}")
print(f"  {line_colour}")
print(f"  {line_fact}")
print(border)
```
