```python
# without_vs_with.py
# ❌ Without conditions — no intelligence
print("Access granted")  # always! even wrong password

# ✅ With conditions — smart decisions
member_name = input("member name")
if member_name == "fathi":
    print("✅ Access granted!")
else:
    print("❌ Wrong password!")

# Real-world examples of conditions:
# - Bank: enough balance? if balance >= amount
# - Age gate: can vote? if age >= 18
# - Grade: A, B, C, D, F based on score
# - Login: username AND password correct?
# - Discount: member? then 20% off
```

```python
# conditions.py
# Basic if / else
age = int(input("Age: "))
if age >= 18:
    print("You can vote! 🗳️")
else:
    print("Not yet — check back later")

# Multiple branches with elif
score = int(input("Score: "))
if   score >= 90: print("A — Excellent!")
elif score >= 80: print("B — Good job!")
elif score >= 70: print("C — Pass")
elif score >= 60: print("D — Borderline")
else:              print("F — Retake")

# Ternary — one-liner condition
status = "adult" if age >= 18 else "minor"
print(f"Status: {status}")
```

```python
# logical.py
# and — BOTH must be True
username = input("Username: ")
password = input("Password: ")
if username == "fathi" and password == "123":
    print("Welcome, admin!") 
else:print("access denied")
# or — EITHER can be True
is_member = True
spending  = 150
if is_member or spending > 100:
    print("10% discount applied!")

# not — flips True to False
is_banned = False
if not is_banned:
    print("You may enter.")

# Combine all three
age   = 20
score = 85
if age >= 18 and score >= 80 and not is_banned:
    print("Eligible for scholarship!")
```

```python
# match_stmt.py
# match — cleaner than many elif
day = input("Day: ").lower()

match day:
    case "monday":
        print("Start of the week!")
    case "friday":
        print("Almost weekend!")
    case "saturday" | "sunday":
        print("Weekend! 🎉")
    case "thursday":
        print("almost friday")
    case _:
        print("A regular weekday")
    

# match with numbers
status = 404
match status:
    case 200: print("OK ✅")
    case 404: print("Not Found 🔍")
    case 500: print("Server Error 💥")
    case _:   print(f"Unknown: {status}")
```
