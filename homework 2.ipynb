```python
# Task 1: Grade Report

name = input("Enter student name: ").strip()
score = float(input("Enter score (0-100): ").strip())

# Determine grade and pass/fail status
if score >= 90:
    grade = "A"
    passed = True
elif score >= 80:
    grade = "B"
    passed = True
elif score >= 70:
    grade = "C"
    passed = True
elif score >= 60:
    grade = "D"
    passed = True
else:
    grade = "F"
    passed = False

status = "Passed" if passed else "Failed"

print("\n--- GRADE REPORT ---")
print(f"Student Name : {name}")
print(f"Score        : {score:.1f}")
print(f"Letter Grade : {grade}")
print(f"Status       : {status}")
```

```python
# Task 2: Smart ATM

PIN = "1234"
balance = 500.0
attempts = 0

authenticated = False

# PIN verification loop (max 3 attempts)
while attempts < 3:
    user_pin = input("Enter 4-digit PIN: ").strip()
    if user_pin == PIN:
        authenticated = True
        print("\nPIN accepted! Welcome to Smart ATM.")
        break
    else:
        attempts += 1
        remaining = 3 - attempts
        if remaining > 0:
            print(f"Incorrect PIN. You have {remaining} attempt(s) remaining.\n")
        else:
            print("\nCard locked! Too many incorrect attempts.")

# Withdrawal process if PIN is correct
if authenticated:
    print(f"Your current balance is: RM{balance:.2f}")
    
    try:
        amount = float(input("Enter withdrawal amount (must be multiple of RM50): ").strip())
        
        if amount <= 0:
            print("Error: Amount must be greater than RM0.")
        elif amount % 50 != 0:
            print("Error: Amount must be a multiple of RM50 (e.g., 50, 100, 150).")
        elif amount > balance:
            print(f"Error: Insufficient funds. Your balance is RM{balance:.2f}.")
        else:
            balance -= amount
            print(f"\nWithdrawal successful! Please collect RM{amount:.2f}.")
            print(f"Remaining balance: RM{balance:.2f}")
    except ValueError:
        print("Error: Invalid input. Please enter a valid numerical amount.")
```

```python
# Task 3: Day Planner

day = input("Enter day of the week (e.g., Monday, Saturday): ").strip().capitalize()
weather = input("Enter weather condition (sunny/rainy/cloudy): ").strip().lower()

# Validate inputs and generate activity plan
if weather not in ["sunny", "rainy", "cloudy"]:
    print("Invalid weather input. Please enter sunny, rainy, or cloudy.")
else:
    match day:
        case "Monday" | "Tuesday" | "Wednesday" | "Thursday" | "Friday":
            if weather == "sunny":
                workout = "Morning Outdoor Run"
                lunch = "Fresh Caesar Salad"
                evening = "Park Walk & Ice Cream"
            elif weather == "rainy":
                workout = "Indoor Yoga & Stretching"
                lunch = "Hot Noodle Soup"
                evening = "Movie Night at Home"
            elif weather == "cloudy":
                workout = "Gym Resistance Training"
                lunch = "Grilled Chicken Wrap"
                evening = "Coffee & Book Reading"

        case "Saturday" | "Sunday":
            if weather == "sunny":
                workout = "Beach Volleyball / Cycling"
                lunch = "Seafood Barbecue"
                evening = "Stargazing or Night Market"
            elif weather == "rainy":
                workout = "Home HIIT Workout"
                lunch = "Comforting Stew"
                evening = "Board Games with Friends"
            elif weather == "cloudy":
                workout = "Hiking Trail Walk"
                lunch = "Café Brunch"
                evening = "Live Music Lounge"
        case _:
            workout, lunch, evening = None, None, None

    if workout:
        print(f"\n--- DAY PLANNER ({day.upper()} - {weather.upper()}) ---")
        print(f"Workout Activity : {workout}")
        print(f"Lunch Suggestion : {lunch}")
        print(f"Evening Activity : {evening}")
    else:
        print("Invalid day entered. Please enter a valid day of the week.")
```

```python
# Task 4: Tax Calculator

income = float(input("Enter your total annual income (RM): ").strip())

tax_total = 0.0
breakdown = []

# Progressive Tax Bands
# Band 1: RM0 - RM5,000 @ 0%
# Band 2: RM5,001 - RM20,000 @ 1% (Max RM15,000 taxable)
# Band 3: RM20,001 - RM35,000 @ 3% (Max RM15,000 taxable)
# Band 4: RM35,001 - RM50,000 @ 8% (Max RM15,000 taxable)
# Band 5: RM50,001 - RM70,000 @ 13% (Max RM20,000 taxable)
# Band 6: RM70,001+ @ 21%

bands = [
    ("RM0 - RM5,000", 5000, 0.00),
    ("RM5,001 - RM20,000", 15000, 0.01),
    ("RM20,001 - RM35,000", 15000, 0.03),
    ("RM35,001 - RM50,000", 15000, 0.08),
    ("RM5,001 - RM70,000", 20000, 0.13),
    ("RM70,001+", float('inf'), 0.21)
]

remaining_income = income

for range_label, band_limit, rate in bands:
    if remaining_income <= 0:
        break
    
    taxable_amount = min(remaining_income, band_limit)
    tax_for_band = taxable_amount * rate
    tax_total += tax_for_band
    
    if taxable_amount > 0:
        breakdown.append({
            "range": range_label,
            "taxable": taxable_amount,
            "rate": f"{int(rate * 100)}%",
            "tax": tax_for_band
        })
        
    remaining_income -= taxable_amount

# Effective Tax Rate (Bonus)
effective_rate = (tax_total / income * 100) if income > 0 else 0.0

# Print Breakdown
print(f"\n--- TAX BREAKDOWN FOR INCOME: RM{income:,.2f} ---")
print(f"{'Tax Band':<22} {'Taxable Amount':>15} {'Rate':>6} {'Tax (RM)':>12}")
print("-" * 58)

for b in breakdown:
    print(f"{b['range']:<22} RM{b['taxable']:>12,.2f} {b['rate']:>6} RM{b['tax']:>9,.2f}")

print("-" * 58)
print(f"Total Tax Payable   : RM{tax_total:,.2f}")
print(f"Effective Tax Rate : {effective_rate:.2f}%")
```
