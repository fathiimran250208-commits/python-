```python
# task1_profile_card.py

# Collect user details
name = input("Enter your name: ").strip()
age = input("Enter your age: ").strip()
country = input("Enter your country: ").strip()
colour = input("Enter your favourite colour: ").strip()
fun_fact = input("Enter a fun fact about yourself: ").strip()

# Prepare content lines
title = " PERSONAL PROFILE "
line_name = f"Name: {name}"
line_age = f"Age: {age}"
line_country = f"Country: {country}"
line_colour = f"Favourite Colour: {colour}"
line_fact = f"Fun Fact: {fun_fact}"

# Calculate dynamic width based on the longest string
max_content_length = max(
    len(title),
    len(line_name),
    len(line_age),
    len(line_country),
    len(line_colour),
    len(line_fact),
)
padding = 10
card_width = max_content_length + padding

border = "=" * card_width

# Output the formatted profile card
print("\n" + border)
print(f"{title:^{card_width}}")
print(border)
print(f"  {line_name}")
print(f"  {line_age}")
print(f"  {line_country}")
print(f"  {line_colour}")
print(f"  {line_fact}")
print(border)
```

```python
# task2_unit_converter.py

# Constants
MILES_PER_KM = 0.621371
LBS_PER_KG = 2.20462

# Inputs
km = float(input("Enter distance in kilometers (km): ").strip())
kg = float(input("Enter weight in kilograms (kg): ").strip())
celsius = float(input("Enter temperature in Celsius (°C): ").strip())

# Conversions
miles = km * MILES_PER_KM
lbs = kg * LBS_PER_KG
fahrenheit = (celsius * 9 / 5) + 32

# Display results formatted to 2 decimal places
print("\n--- Conversion Results ---")
print(f"{km:.2f} km  = {miles:.2f} miles")
print(f"{kg:.2f} kg  = {lbs:.2f} lbs")
print(f"{celsius:.2f} °C = {fahrenheit:.2f} °F")
```

```python
# task3_receipt_calculator.py

items = []

# Collect input for 3 items
print("Please enter details for 3 items:\n")
for i in range(1, 4):
    print(f"Item {i}:")
    name = input("  Name: ").strip()
    price = float(input("  Price ($): ").strip())
    quantity = int(input("  Quantity: ").strip())
    total_price = price * quantity
    items.append({"name": name, "price": price, "qty": quantity, "total": total_price})
    print()

# Calculations
subtotal = sum(item["total"] for item in items)
gst = subtotal * 0.06
service_charge = subtotal * 0.10
grand_total = subtotal + gst + service_charge

# Print Receipt
width = 45
print("=" * width)
print(f"{'CAFE RECEIPT':^{width}}")
print("=" * width)
print(f"{'Item':<18} {'Qty':>5} {'Price ($)':>10} {'Total ($)':>10}")
print("-" * width)

for item in items:
    print(
        f"{item['name']:<18} {item['qty']:>5} {item['price']:>10.2f} {item['total']:>10.2f}"
    )

print("-" * width)
print(f"{'Subtotal:':<34} ${subtotal:>8.2f}")
print(f"{'GST (6%):':<34} ${gst:>8.2f}")
print(f"{'Service Charge (10%):':<34} ${service_charge:>8.2f}")
print("=" * width)
print(f"{'Grand Total:':<34} ${grand_total:>8.2f}")
print("=" * width)
```

```python
# task4_string_inspector.py

sentence = input("Enter a sentence: ").strip()

length = len(sentence)
words = sentence.split()
word_count = len(words)
upper_case = sentence.upper()
title_case = sentence.title()
contains_python = "python" in sentence.lower()
reversed_sentence = sentence[::-1]

print("\n--- String Inspection Report ---")
print(f"Original:          {sentence}")
print(f"Character Count:   {length}")
print(f"Word Count:        {word_count}")
print(f"Uppercase:         {upper_case}")
print(f"Title Case:        {title_case}")
print(f"Contains 'python': {contains_python}")
print(f"Reversed:          {reversed_sentence}")
```
