```python

# error_handling.py
# ❌ Crash if file missing
# f = open("data.txt")   # FileNotFoundError!

# ✅ Handle it gracefully
try:
    f = open("data.txt")
    content = f.read()
except FileNotFoundError:
    print("File not found!")
except PermissionError:
    print("No permission to read file")
finally:
    print("This always runs")

# Catching multiple + custom message
try:
    age = int(input("Enter age: "))
    if age < 0: raise ValueError("Age cannot be negative")
except ValueError as e:
    print(f"Error: {e}")

# Custom exception
class InsufficientFundsError(Exception):
    pass

raise InsufficientFundsError("Balance too low!")
```

```python
# file_json.py
# Reading a text file
with open("notes.txt", "r") as f:
    content = f.read()
    # or: lines = f.readlines()

# Writing a text file
with open("output.txt", "w") as f:
    f.write("Hello world!")

# Appending to file
with open("log.txt", "a") as f:
    f.write("New log entry")

# JSON — save Python dict to file
import json

data = {"name": "Ali", "scores": [85, 92, 78]}

with open("data.json", "w") as f:
    json.dump(data, f, indent=2)

# Load JSON from file
with open("data.json") as f:
    loaded = json.load(f)

print(loaded["name"])     # Ali
print(loaded["scores"])   # [85, 92, 78]
```
