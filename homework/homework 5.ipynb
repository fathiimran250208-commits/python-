```python

class Contact:
    def __init__(self, name, phone, email, category):
        self.name = name
        self.phone = phone
        self.email = email
        self.category = category

    def display(self):
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")
        print(f"Email: {self.email}")
        print(f"Category: {self.category}")

    def update(self, field, value):
        if field in ["name", "phone", "email", "category"]:
            setattr(self, field, value)
            return True
        return False

    def matches(self, query):
        query = query.lower()
        return (
            query in self.name.lower()
            or query in self.phone.lower()
            or query in self.email.lower()
            or query in self.category.lower()
        )


class ContactBook:
    def __init__(self):
        self.contacts = []

    def add(self, contact):
        self.contacts.append(contact)

    def remove(self, name):
        for contact in self.contacts:
            if contact.name.lower() == name.lower():
                self.contacts.remove(contact)
                print("Contact removed.")
                return
        print("Contact not found.")

    def search(self, query):
        results = [c for c in self.contacts if c.matches(query)]
        if results:
            for contact in results:
                contact.display()
                print()
        else:
            print("No contacts found.")

    def list_by_category(self, category):
        for contact in self.contacts:
            if contact.category.lower() == category.lower():
                contact.display()
                print()


# Example program
book = ContactBook()

book.add(Contact("Ali", "12345678", "ali@example.com", "Friend"))
book.add(Contact("Sara", "87654321", "sara@example.com", "Family"))

print("ALL CONTACTS")
book.search("")

print("SEARCH FOR ALI")
book.search("Ali")

print("FRIENDS")
book.list_by_category("Friend")

print("UPDATE ALI'S PHONE")
book.contacts[0].update("phone", "99999999")
book.contacts[0].display()

print("REMOVE SARA")
book.remove("Sara")
```

```python

class Vehicle:
    def __init__(self, make, model, year, mileage):
        self.make = make
        self.model = model
        self.year = year
        self.mileage = mileage

    def drive(self, km):
        self.mileage += km
        print(f"{self.make} {self.model} drove {km} km.")
        print(f"Total mileage: {self.mileage} km")


class Car(Vehicle):
    def __init__(self, make, model, year, mileage, doors=4):
        super().__init__(make, model, year, mileage)
        self.doors = doors

    def drive(self, km):
        super().drive(km)
        print(f"Number of doors: {self.doors}")


class Truck(Vehicle):
    def __init__(self, make, model, year, mileage, cargo_kg):
        super().__init__(make, model, year, mileage)
        self.cargo_kg = cargo_kg

    def drive(self, km):
        super().drive(km)
        self.cargo_kg = max(0, self.cargo_kg - km)
        print(f"Remaining cargo capacity: {self.cargo_kg} kg")


class ElectricCar(Vehicle):
    def __init__(self, make, model, year, mileage,
                 battery_kwh, range_km):
        super().__init__(make, model, year, mileage)
        self.battery_kwh = battery_kwh
        self.range_km = range_km

    def drive(self, km):
        super().drive(km)
        self.range_km = max(0, self.range_km - km)
        print(f"Remaining range: {self.range_km} km")


# Example program
car = Car("Toyota", "Corolla", 2022, 15000)
truck = Truck("Ford", "F-150", 2021, 30000, 1000)
electric = ElectricCar("Tesla", "Model 3", 2024, 5000, 75, 400)

print("=== CAR ===")
car.drive(50)

print("\n=== TRUCK ===")
truck.drive(100)

print("\n=== ELECTRIC CAR ===")
electric.drive(80)
```

```python

from dataclasses import dataclass
from datetime import date, timedelta


@dataclass
class Book:
    title: str
    author: str
    isbn: str
    year: int
    available: bool = True
    checkout_date: date = None


class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)
        print(f"Added: {book.title}")

    def checkout(self, isbn):
        for book in self.books:
            if book.isbn == isbn:
                if book.available:
                    book.available = False
                    book.checkout_date = date.today()
                    print(f"Checked out: {book.title}")
                else:
                    print("Book is already checked out.")
                return
        print("Book not found.")

    def return_book(self, isbn):
        for book in self.books:
            if book.isbn == isbn:
                if not book.available:
                    book.available = True
                    book.checkout_date = None
                    print(f"Returned: {book.title}")
                else:
                    print("Book is already available.")
                return
        print("Book not found.")

    def search(self, query):
        query = query.lower()
        found = False

        for book in self.books:
            if (query in book.title.lower()
                    or query in book.author.lower()
                    or query in book.isbn.lower()):
                status = "Available" if book.available else "Checked out"
                print(f"{book.title} by {book.author} - {status}")
                found = True

        if not found:
            print("No books found.")

    def overdue_books(self):
        today = date.today()
        found = False

        for book in self.books:
            if not book.available and book.checkout_date:
                days = (today - book.checkout_date).days

                if days >= 14:
                    print(f"{book.title}: {days} days overdue-check")
                    found = True

        if not found:
            print("No overdue books.")

    def display_books(self):
        for book in self.books:
            status = "Available" if book.available else "Checked out"
            print(f"{book.title} | {book.author} | {status}")


# Example program
library = Library()

book1 = Book("Harry Potter", "J.K. Rowling", "101", 1997)
book2 = Book("The Hobbit", "J.R.R. Tolkien", "102", 1937)
book3 = Book("Matilda", "Roald Dahl", "103", 1988)

library.add_book(book1)
library.add_book(book2)
library.add_book(book3)

print("\n=== CHECKOUT ===")
library.checkout("101")
library.checkout("102")

# Simulate a book checked out 20 days ago
book2.checkout_date = date.today() - timedelta(days=20)

print("\n=== SEARCH ===")
library.search("Harry")

print("\n=== ALL BOOKS ===")
library.display_books()

print("\n=== OVERDUE BOOKS ===")
library.overdue_books()

print("\n=== RETURN BOOK ===")
library.return_book("101")

print("\n=== CHECK AGAIN ===")
library.display_books()
```

```python

class Character:
    def __init__(self, name, hp, attack, defence):
        self.name = name
        self.hp = hp
        self.attack_power = attack
        self.defence = defence

    def attack_target(self, other):
        damage = max(0, self.attack_power - other.defence)
        other.hp = max(0, other.hp - damage)
        print(f"{self.name} attacks {other.name} for {damage} damage!")

    def defend(self):
        print(f"{self.name} is defending!")

    def is_alive(self):
        return self.hp > 0

    def __repr__(self):
        return f"{self.name}: HP={self.hp}"


class Warrior(Character):
    def __init__(self, name, hp, attack, defence, shield_bonus):
        super().__init__(name, hp, attack, defence)
        self.shield_bonus = shield_bonus

    def defend(self):
        self.defence += self.shield_bonus
        print(f"{self.name} raises a shield! Defence is now {self.defence}.")

    def __repr__(self):
        return f"Warrior({self.name}, HP={self.hp}, Defence={self.defence})"


class Mage(Character):
    def __init__(self, name, hp, attack, defence, mana, spells):
        super().__init__(name, hp, attack, defence)
        self.mana = mana
        self.spells = spells

    def attack_target(self, other):
        if self.mana >= 10:
            self.mana -= 10
            damage = max(0, self.attack_power + 10 - other.defence)
            other.hp = max(0, other.hp - damage)
            print(f"{self.name} casts a spell on {other.name} "
                  f"for {damage} damage!")
        else:
            super().attack_target(other)

    def __repr__(self):
        return f"Mage({self.name}, HP={self.hp}, Mana={self.mana})"


class Rogue(Character):
    def __init__(self, name, hp, attack, defence, stealth):
        super().__init__(name, hp, attack, defence)
        self.stealth = stealth

    def attack_target(self, other):
        bonus = 5 if self.stealth else 0
        damage = max(0, self.attack_power + bonus - other.defence)
        other.hp = max(0, other.hp - damage)
        print(f"{self.name} attacks {other.name} for {damage} damage!")
        if self.stealth:
            self.stealth = False

    def __repr__(self):
        return f"Rogue({self.name}, HP={self.hp}, Stealth={self.stealth})"


class Battle:
    def __init__(self, character1, character2):
        self.character1 = character1
        self.character2 = character2
        self.log = []

    def fight(self):
        for round_number in range(1, 4):
            if not (self.character1.is_alive()
                    and self.character2.is_alive()):
                break

            message = f"\n--- ROUND {round_number} ---"
            self.log.append(message)
            print(message)

            self.character1.attack_target(self.character2)
            self.log.append(
                f"{self.character1.name} HP={self.character1.hp}; "
                f"{self.character2.name} HP={self.character2.hp}"
            )

            if self.character2.is_alive():
                self.character2.attack_target(self.character1)
                self.log.append(
                    f"{self.character1.name} HP={self.character1.hp}; "
                    f"{self.character2.name} HP={self.character2.hp}"
                )

            print(self.character1)
            print(self.character2)

        print("\n=== BATTLE RESULT ===")
        if self.character1.is_alive() and not self.character2.is_alive():
            print(f"{self.character1.name} wins!")
        elif self.character2.is_alive() and not self.character1.is_alive():
            print(f"{self.character2.name} wins!")
        elif self.character1.is_alive() and self.character2.is_alive():
            print("Both characters survived!")
        else:
            print("Both characters were defeated!")

        print("\n=== BATTLE LOG ===")
        for entry in self.log:
            print(entry)


# Example program
warrior = Warrior("Knight", 100, 25, 10, 5)
mage = Mage("Wizard", 90, 30, 5, 50, ["Fireball", "Ice"])

battle = Battle(warrior, mage)
battle.fight()
```
